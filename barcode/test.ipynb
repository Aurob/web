```python
import cv2
from pyzbar.pyzbar import decode

def main():
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        raise RuntimeError("Cannot open camera")

    seen = set()
    print("Press Ctrl+C to quit. Hold barcode steady in frame.")

    try:
        while True:
            ret, frame = cap.read()
            if not ret:
                break

            barcodes = decode(frame)
            for b in barcodes:
                data = b.data.decode("utf-8", errors="replace")
                kind = b.type
                key = (kind, data)

                if key not in seen:
                    seen.add(key)
                    print(f"\nTYPE: {kind}")
                    print(f"VALUE: {data}")
                    print(f"LENGTH: {len(data)}")

                # draw box for feedback
                x, y, w, h = b.rect
                cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
                cv2.putText(
                    frame,
                    kind,
                    (x, y - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6,
                    (0, 255, 0),
                    2,
                )

            cv2.imshow("Barcode Scanner", frame)
            if cv2.waitKey(1) & 0xFF == 27:  # ESC
                break

    finally:
        cap.release()
        cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
```
