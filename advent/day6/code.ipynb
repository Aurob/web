```python
# Advent of Code Day 6
# https://adventofcode.com/2025/day/6

import re
from pprint import pprint
def day6_part1(input_str):
    _rows = input_str.split('\n')

    col_rows = []
    for r, row in enumerate(_rows):
        if not row: continue
        col_vals = re.findall('\d+|[\*|\+]+', row)

        col_rows.append([])
        for col in range(len(col_vals)):
            col_rows[len(col_rows)-1].append(col_vals[col])
            
    col_set = []
    for r in col_rows:
        s = []
        for c in range(len(r)):
            s.append(r[c])
        col_set.append(s)
    
    rotated = []
    for row in range(len(col_set[0])):
        col_vals = []
        for r in range(len(col_set)-1):
            col_vals.append(col_set[r][row])
        rotated.append(col_vals)
        

    op_results = []
    for i in range(len(rotated)):
        col = rotated[i]
        op = col[-1]
        val = 0
        match op:
            case '*':
                val = int(col[0])
                for r in range(1, len(col)-1):
                    val *= int(col[r])
            case '+':
                val = int(col[0])
                for r in range(1, len(col)-1):
                    val += int(col[r])
                pass
            case _:
                pass
        
        op_results.append(val)

    return op_results

def test():

    input_test = '''
    123 328  51 64
    45 64  387 23
    6 98  215 314
    *   +   *   +
    '''
    '''
    123 * 45 * 6 = 33210
    328 + 64 + 98 = 490
    51 * 387 * 215 = 4243455
    64 + 23 + 314 = 401
    '''

    return day6_part1(input_test)

def final():
    with open("input/input.txt", "r") as f:
        input_file = f.read()

        print(input_file)
        part1 = day6_part1(input_file)
        return part1

pprint(sum(final()))
# pprint(sum(test()))
```
