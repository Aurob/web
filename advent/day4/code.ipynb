```python
# Advent of Code Day 4
# https://adventofcode.com/2025/day/4


def day4_part1(input_str):

    def find_forklift_spots(grid):
        grid_size = len(grid), len(grid[0])

        cell_neighbors = {}

        x_count = 0

        for row in range(grid_size[0]):
            # if row != 0: continue

            for col in range(grid_size[1]):
                # if col != 0: continue

                cell = grid[row][col]

                nol = row == 0
                nor = row == grid_size[0]-1
                noa = col == 0
                nob = col == grid_size[1]-1

                neighbors = [
                    grid[row-1][col-1] if not (nol or noa) else None, 
                    grid[row][col-1] if not (noa) else None, 
                    grid[row+1][col-1] if not (nor or noa) else None,
                    grid[row-1][col] if not (nol) else None,
                    grid[row+1][col] if not (nor) else None,
                    grid[row-1][col+1] if not (nol or nob) else None, 
                    grid[row][col+1] if not (nob) else None, 
                    grid[row+1][col+1] if not (nor or nob) else None,
                ]

                # print(neighbors, [n for n in neighbors if n == '@'])
                if cell == '@' and len([n for n in neighbors if n == '@']) < 4:
                    # make_x.append((row, col))
                    x_count += 1
        
        # for cell in make_x:
        #     grid[cell[0]][cell[1]] = 'x'

        return x_count #grid, cell_neighbors

    grid = []
    for line in input_str.split('\n'):
        line = line.strip()
        if not line: continue

        row = [c for c in line.strip()]
        grid.append(row)

    x_count = find_forklift_spots(grid)
    
    return x_count

def test():
    # From the challenge
    input_test = '''
        ..@@.@@@@.
        @@@.@.@.@@
        @@@@@.@.@@
        @.@@@@..@.
        @@.@@@@.@@
        .@@@@@@@.@
        .@.@.@.@@@
        @.@@@.@@@@
        .@@@@@@@@.
        @.@.@@@.@.
    '''

    input_solution = '''
        ..xx.xx@x.
        x@@.@.@.@@
        @@@@@.x.@@
        @.@@@@..@.
        x@.@@@@.@x
        .@@@@@@@.@
        .@.@.@.@@@
        x.@@@.@@@@
        .@@@@@@@@.
        x.x.@@@.x.
    '''

    part1 = day4_part1(input_test)
    # return ''.join([''.join(row) for row in part1]) == input_solution.replace(' ', '').replace('\n', '')
    return part1

def final():
    with open("input.txt", "r") as f:
        input_file = f.read()

        part1 = day4_part1(input_file)
        return part1

print(final())
```
