```python
# Advent of Code Day 7
# https://adventofcode.com/2025/day/7

import re
from pprint import pprint
import random as rn

def day7_part1(input_str):
    lines = []

    split_map = {}
    split_count = 0

    # get a clean list of lines 
    for l, line in enumerate(input_str.split('\n')):
        line = line.strip()
        if not line: continue
        lines.append(line)

        # pre-populate the split_map with all split character positions
        for i, char in enumerate(line):
            if char == '^':
                split_map[(len(lines)-1, i)] = {
                    'source': (None, None),
                    'left': (None, None),
                    'right': (None, None)
                }

    _lines = []
    start = lines[0].index('S')
    for l, line in enumerate(lines):

        _line = list(line)

        # The first line will always be unchangd
        if l > 0:
            # if on the second line, there will always be 1 beam under the start index
            if l == 1:
                _line[start] = '|'

            # if the current character is "^"
            for i, char in enumerate(line):
                prev_line_val = _lines[l-1][i]

                if char == '^':
                    # if the same index on the previous line is a beam
                    # then set the source, left, and right beam positions in the split_map
                    if prev_line_val == '|':
                        split_count += 1

                        split_map[(l, i)]['source'] = (l-1, i)
                        split_map[(l, i)]['left'] = (l+1, i-1)
                        split_map[(l, i)]['right'] = (l+1, i+1)



            # if the current character is "."
            for i, char in enumerate(line):
                prev_line_val = _lines[l-1][i]
                if char == '.':
                    
                    # if the same index on the previous line is a beam
                    # then this character becomes a beam
                    if prev_line_val == '|':
                        _line[i] = '|'

                    # if the value at index-1 or index+1 on the previous line is a splitter
                    # and it has a source beam
                    # then this character becomes a beam
                    if i > 0:
                        if _line[i-1] == '^':
                            left_split = split_map[(l, i-1)]
                            if left_split['source'] != (None, None):
                                _line[i] = '|'

                    if i < len(_line)-1:
                        if _line[i+1] == '^':
                            right_split = split_map[(l, i+1)]
                            if right_split['source'] != (None, None):
                                _line[i] = '|'

        _lines.append(''.join(_line))

    return split_count

def day7_part2(input_str):
    lines = []

    split_map = {}
    split_count = 0

    # get a clean list of lines 
    for l, line in enumerate(input_str.split('\n')):
        line = line.strip()
        if not line: continue
        lines.append(line)

        # pre-populate the split_map with all split character positions
        for i, char in enumerate(line):
            if char == '^':
                split_map[(len(lines)-1, i)] = {
                    'source': (None, None),
                    'left': (None, None),
                    'right': (None, None)
                }

    left_only = True

    _lines = []
    start = lines[0].index('S')
    for l, line in enumerate(lines):

        _line = list(line)

        # The first line will always be unchangd
        if l > 0:
            # if on the second line, there will always be 1 beam under the start index
            if l == 1:
                _line[start] = '|'

            # if the current character is "^"
            for i, char in enumerate(line):
                prev_line_val = _lines[l-1][i]

                if char == '^':
                    # if the same index on the previous line is a beam
                    # then set the source, left, and right beam positions in the split_map
                    if prev_line_val == '|':
                        split_count += 1

                        split_map[(l, i)]['source'] = (l-1, i)
                        split_map[(l, i)]['left'] = (l+1, i-1)
                        split_map[(l, i)]['right'] = (l+1, i+1)



            # if the current character is "."
            for i, char in enumerate(line):
                prev_line_val = _lines[l-1][i]
                if char == '.':
                    
                    # if the same index on the previous line is a beam
                    # then this character becomes a beam
                    if prev_line_val == '|':
                        _line[i] = '|'

                    # if the value at index-1 or index+1 on the previous line is a splitter
                    # and it has a source beam
                    # then this character becomes a beam

                    new_beams = []
                    if i > 0:
                        if _line[i-1] == '^':
                            left_split = split_map[(l, i-1)]
                            if left_split['source'] != (None, None):
                                # _line[i] = '|'
                                new_beams.append(i)

                    if i < len(_line)-1:
                        if _line[i+1] == '^':
                            right_split = split_map[(l, i+1)]
                            if right_split['source'] != (None, None):
                                # _line[i] = '|'
                                new_beams.append(i)

                    print(len(new_beams))
                    if len(new_beams) == 2:
                        # if rn.randint(0, 1):
                        _line[new_beams[1]] = '|'
                        # else:
                        #     _line[new_beams[0]] = '|'
                    elif new_beams:
                        _line[new_beams[0]] = '|'

        _lines.append(''.join(_line))
    print('\n'.join(_lines))

    return split_count

def test():

    input_test = '''
    .......S.......
    ...............
    .......^.......
    ...............
    ......^.^......
    ...............
    .....^.^.^.....
    ...............
    ....^.^...^....
    ...............
    ...^.^...^.^...
    ...............
    ..^...^.....^..
    ...............
    .^.^.^.^.^...^.
    ...............
    '''

    test_solution = '''
    .......S.......
    .......|.......
    ......|^|......
    ......|.|......
    .....|^|^|.....
    .....|.|.|.....
    ....|^|^|^|....
    ....|.|.|.|....
    ...|^|^|||^|...
    ...|.|.|||.|...
    ..|^|^|||^|^|..
    ..|.|.|||.|.|..
    .|^|||^||.||^|.
    .|.|||.||.||.|.
    |^|^|^|^|^|||^|
    |.|.|.|.|.|||.|
    '''
    return day7_part2(input_test)

def final():
    with open("input/input.txt", "r") as f:
        input_file = f.read()

        # print(input_file)
        part1 = day7_part1(input_file)
        return part1

# pprint(sum(final()))
pprint(test())
```
