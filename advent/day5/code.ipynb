```python
# Advent of Code Day 5
# https://adventofcode.com/2025/day/5

def day5_part1(input_str):

    blank_line = input_str.index('\n\n')
    fresh_ids = list(set(input_str[:blank_line].strip().replace('\n', ',').replace(' ', '').split(',')))
    ingredients = input_str[blank_line:].strip().replace('\n', ',').replace(' ', '').split(',')
    fresh_set = set()
    ingredients = set([int(i) for i in ingredients])
    
    for i, val in enumerate(fresh_ids):
        dash = val.index('-')
        low = int(val[:dash])
        high = int(val[dash+1:])
        fresh_ids[i] = (low, high)


    for ingrid in ingredients:
        for fresh in fresh_ids:
            low, high = fresh
            if ingrid >= low and ingrid <= high:    
                fresh_set.add(ingrid)

    fresh_ingredients = ingredients.intersection(fresh_set)

    return len(fresh_ingredients)

def day5_part2(input_str):

    blank_line = input_str.index('\n\n')
    base_fresh_ids = list(set(input_str[:blank_line].strip().replace('\n', ',').replace(' ', '').split(',')))
    ingredients = input_str[blank_line:].strip().replace('\n', ',').replace(' ', '').split(',')
    fresh_set = set()

    ingredients = set([int(i) for i in ingredients])
    lowest = None
    highest = 0
    lows = []
    highs = []
    fresh_sets = []
    fresh_ids = base_fresh_ids
    for i, val in enumerate(base_fresh_ids):
        low, high = val.split('-')
        lows.append(int(low))
        highs.append(int(high))
        fresh_ids[i] = (lows[i], highs[i])

    # sort the ranges by the sum of the number of ids in the range
    # a = list(map(lambda x: sum(x), fresh_ids))
    # fresh_ids = [fresh_ids[a.index(i)] for i in sorted(a)]
    # print(base_fresh_ids, fresh_ids)

    # check for ranges where the high is equal to the low in another
    # these can be merged into a single range

    def merge_ranges(ranges):
        skip = []
        _ranges = ranges
        ranges_ = []
        checked = set()
        for i, ilowhigh in enumerate(ranges):
            if i in skip: continue
            il,ih = ilowhigh
            had_merge = False
            ilih = [il, ih]
            for j, jlowhigh in enumerate(ranges):
                if j == i: continue
                if j in skip: continue
                jl, jh = jlowhigh
                _il, _ih = ilih

                # if (f'{(jlowhigh, (_il, _ih))}') in checked:
                #     continue
                # checked.add(f'{(jlowhigh, (_il, _ih))}')

                # Main: 5-10
                #  - iter: 
                #   - 5-10 # iter low is equal to main low AND iter high is equal to main high
                #       - ignore
                if jl == _il and _ih == jh:
                    had_merge = True
                    skip.append(j)
                    continue

                #   - 6-9 # iter low is higher than main low AND iter high is lower than main high
                #       - ignore
                elif jl > _il and jh < _ih:                    
                    had_merge = True
                    skip.append(j)
                    continue

                #   - 4-7 # iter low is lower than main low AND iter high is higher than the main low
                #       - main low becomes iter low
                elif jl < _il and jh > _il:
                    ilih[0] = jl

                #   - 5-11 # iter low is equal to main high AND iter high is higher than main high
                #       - main high becomes iter high
                elif jl == _ih and jh > _ih:
                    ilih[1] = jh
                elif jl == _il and jh > _ih:
                    ilih[1] = jh
                
                #   - 4-10 # iter low is lower than main low AND iter high is equal to main high
                #       - main low becomes iter low
                elif jl < _il and jh == _ih:
                    ilih[0] = jl

                # (6, 9), (5, 10)]
                elif _il > jl and _ih < jh:
                    ilih[0] = jl
                    ilih[1] = jh

                #   - 4-11 # iter low is lower than main low AND iter high is higher than main high
                #       - main low becomes iter low and main high becomes iter high
                elif jl < _il and jh > _ih:
                    ilih[0] = jl
                    ilih[1] = jh
                # else:
                #     print(_il, ih, jl, jh)

                if [_il, _ih] != ilih:
                    had_merge = True
                    skip.append(j)

            ranges_final = []
            for r, rang in enumerate(ranges):
                if r == i:
                    rang = ilih
                if r not in skip:
                    ranges_final.append(rang)
                
            ranges = ranges_final
            if had_merge:
                break

        return ranges, had_merge

    fresh_ids, had_merge = merge_ranges(fresh_ids)
    i = 0
    while had_merge:

        fresh_ids, had_merge= merge_ranges(fresh_ids)
        i += 1

    a = list(map(lambda x: sum(x), fresh_ids))
    b = sorted(a)
    merged_fresh_ids = [fresh_ids[a.index(i)] for i in b]
    range_totals = [((fi[1]-fi[0])+1) for fi in merged_fresh_ids]
    total_fresh_ids = sum(range_totals)

    return total_fresh_ids


def test():

    # X X 3 4 5 X X X X 10 11 12
    
    input_test = '''
    3-5
    10-14
    16-20
    12-18

    1
    5
    8
    11
    17
    32
    '''


    i0='''
    5-10
    5-10

    1
    '''
    i1='''
    5-10
    4-7

    1
    '''
    i2='''
    5-10
    5-11

    1
    '''
    i3='''
    5-10
    4-10

    1
    '''
    i4='''
    5-10
    4-11

    1
    '''
    i5='''
    5-10
    6-9

    1
    '''

    input_solution = '''
    '''
    # for i in [i0, i1, i2, i3, i4, i5]:
    #     print(day5_part2(i))
    print(day5_part2(input_test))

def final():
    with open("input/input.txt", "r") as f:
        input_file = f.read()

        part1 = day5_part2(input_file)
        return part1

print(final())
#359132249205590
#359490987293418
#368685600348275
#323779396663803
#133015481316888
#395077814563647
#332998283036769
# print(test())
```
