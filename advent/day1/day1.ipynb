```python
# Advent of Code Day 1
# https://adventofcode.com/2025/day/1


def day1(input_str:str):
    position = 50
    zero_states = 0
    zero_clicks = 0

    input_data = [l.strip().replace(' ', '') for l in input_str.replace('\n', ' ').split(' ')]

    for state in input_data:
        if not state: continue
        direction = -1 if state[0] == 'L' else 1
        clicks = int(state[1:])

        for c in range(clicks):
            position += direction
            if position == 100:
                position = 0
            if position == -1:
                position = 99
            
            if position == 0:
                zero_clicks += 1

        if position == 0:
            zero_states += 1

    return zero_states, zero_clicks

def test():

    # From the challenge
    input_test = '''
    L68
    L30
    R48
    L5
    R60
    L55
    L1
    L99
    R14
    L82'''

    part1, part2 = day1(input_test)
    assert part1 == 3
    assert part2 == 6


with open("input.txt", "r") as f:
    input_file = f.read()
    
    print(day1(input_file))
```
