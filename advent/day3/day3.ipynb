```python
# Advent of Code Day 3
# https://adventofcode.com/2025/day/3

def day3_part1(input_str):
    line_joltages = []

    for line in input_str.split('\n'):
        if not line: continue
        nums = list(line.strip())
        max_combo = 0
        for i, num in enumerate(nums):
            for j, n in enumerate(nums):
                if j == i: continue
                if j < i: continue
                combo = int(num+n)
                if combo > max_combo:
                    max_combo = combo

        line_joltages.append(max_combo)

    return line_joltages

def day3_part2(input_str):
    line_joltages = []

    def get_max(nums, limit=12, p=0):
        leading_digit = 0
        leading_index = 0

        for i, n in enumerate(nums):
            num = int(n)
            if num > leading_digit:
                if p:
                    print(num, leading_digit, i, len(nums) - i)
                if len(nums) - i >= limit:
                    leading_digit = num
                    leading_index = i

        return leading_index, leading_digit

  
    def process_nums(nums, limit):
        first_index, first_digit = get_max(nums, limit)
        available_digits = nums[first_index+1:]
        return first_digit, available_digits
    
    
    for line in input_str.split('\n'):
        if not line: continue
        nums = line.strip()

        print(nums)

        limit = 12
        stack = []
        while len(stack) < 12:
            first_digit, nums = process_nums(nums, limit)
            stack.append(first_digit)
            limit -=1
        joltage = int(''.join([str(i) for i in stack]))
        print(joltage)
        line_joltages.append(joltage)

    return line_joltages
```
